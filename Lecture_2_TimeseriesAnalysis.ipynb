```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller
```

```python
df =pd.read_csv(
    'https://raw.githubusercontent.com/jbrownlee/Datasets/master/monthly-mean-temp.csv',
    parse_dates=True,
    index_col=0
)
```

```python
df.head(10)
```

```python
df.plot()
```

```python
#testing for stationarity
result = adfuller(df)
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
```

```python
#since the p-value is above the set threshold of 0.05, the data is not stationary
```
