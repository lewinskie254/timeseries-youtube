```python
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller
```

```python
df =pd.read_csv(
    'https://raw.githubusercontent.com/jbrownlee/Datasets/master/monthly-mean-temp.csv',
    parse_dates=True,
    index_col=0
)
```

```python
df.head(10)
```

```python
df.plot()
```

```python
#testing for stationarity
result = adfuller(df)
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
```

```python
#since the p-value is below the set threshold of 0.05, the data is stationary
```

```python
df2 = pd.read_csv("https://raw.githubusercontent.com/statsmodels/smdatasets/refs/heads/main/data/stl-decomposition/co2.csv",
                  parse_dates=True,
                  index_col=0)
```

```python
df2.plot()
```

```python
result2 = adfuller(df2)
print('ADF Statistic: %f' % result2[0])
print('p-value: %f' % result2[1])
```

```python
#differentiate

diff_co2 = df2.diff()
```

```python
diff_co2.plot()
```

```python
#start from the second element
result3 = adfuller(diff_co2[1:])
print('ADF Statistic: %f' % result3[0])
print('p-value: %f' % result3[1])
```
