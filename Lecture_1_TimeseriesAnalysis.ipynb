```python
import matplotlib.pyplot as plt
import pandas as pd
from pandas.plotting import register_matplotlib_converters
import seaborn as sns
```

```python
co2 = pd.read_csv("https://raw.githubusercontent.com/statsmodels/smdatasets/refs/heads/main/data/stl-decomposition/co2.csv", parse_dates=True,
    index_col=0)
```

```python
co2.head(10)
```

```python
co2.plot()
```

```python
from statsmodels.tsa.seasonal import STL

stl = STL(co2, seasonal=13)
res = stl.fit()
fig = res.plot()
```
